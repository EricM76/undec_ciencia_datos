En Python, df es la abreviatura convencional para un DataFrame de Pandas, que es la estructura de datos fundamental de la biblioteca Pandas, diseñada para manipular y analizar datos tabulares bidimensionales (filas y columnas).
Funciona similar a una hoja de cálculo o una tabla SQL, permitiendo gestionar datos heterogéneos.
Aquí los detalles clave sobre df:

¿Qué es?: Un DataFrame de Pandas, un objeto tabular de dos dimensiones.

Convención: Aunque se puede llamar de cualquier forma, usar df es un estándar aceptado entre analistas y científicos de datos para identificar rápidamente un DataFrame.

Estructura: Se compone de filas etiquetadas y columnas que pueden contener diferentes tipos de datos (números, texto, booleanos).

```python
# Con esta línea los que hacemos es tomar la columna 'salary' que contiene los valores del salario en formato de cadena,
# eliminamos cualquier caracter que no sea numérico, extraemos los dígitos y convertimos la columna resultante en valores numéricos de tipo float.
# El valor resultante se almacena en una nueva columna llamada 'salary_numeric'.
# Esto es muy útil cuando la columna 'salary' inicial contiene caracteres no numéricos y se desea trabajar con los salarios como valores numéricos.

df_sin_nulos['salary_numeric'] = df_sin_nulos['salary'].replace('[^\d.]', '', regex=True).str.extract('(\d+)').astype(float)
```

# **6° Paso: Reemplazamos los Valores Nulos con la Media**

6_1 Rellenamos

```python
# Reemplazamos los valores nulos con un valor específico, en este caso el promedio

df_con_reemplazo = df_sin_nulos.fillna(df_sin_nulos['salary_numeric'].mean())
```

```python
df_con_reemplazo
```

```python
df_con_reemplazo.drop('salary', axis=1, inplace=True)
```

# **7° Paso: Agrupamos y Agregamos Datos**

```python
# Continuando con el ejemplo tomamos las columnas 'title' y 'salary'.
# Vamos a agrupar por título y calcular el promedio y la mediana del salario para cada título.

df_grouped = df_con_reemplazo.groupby('title')['salary_numeric'].agg(['mean', 'median', 'count']).reset_index()

# Renombramos las columnas para que se entienda mejor
df_grouped.columns = ['title', 'salary_mean', 'salary_median', 'job_count']

# Finalmente mostramos el nuevo DataFrame resultante
df_grouped
```

```python
# Sacamos la Media
df_grouped['salary_mean'] = df_grouped['salary_mean'].round(0).astype(int)

# Sacamos la Mediana
df_grouped['salary_median'] = df_grouped['salary_median'].round(0).astype(int)
```

```python
# Mostramos el nuevo DataSet
df_grouped
```

# **2° Paso: Identificación de Datos Inconsistentes y Outliers**

```python
# Ahora vamos a detectar valores lógicos que están fuera de rango
datos_inconsistentes = df_Lavoral[
    (df_Lavoral['Role_Relevance'] < 0) | (df_Lavoral['Role_Relevance'] > 10)
]

print("Registros con datos inconsistentes:", len(datos_inconsistentes))
```

```python
# detección de datos Outliers (IQR)
def detectar_outliers_iqr(df_Lavoral, col):
    Q1 = df_Lavoral[col].quantile(0.25)
    Q3 = df_Lavoral[col].quantile(0.75)
    IQR = Q3 - Q1
    return df_Lavoral[(df_Lavoral[col] < Q1 - 1.5 * IQR) | (df_Lavoral[col] > Q3 + 1.5 * IQR)]

outliers_role = detectar_outliers_iqr(df_Lavoral, 'Role_Relevance')
print("Outliers en Role_Relevance:", len(outliers_role))
```

```python
# Vamos a utilizar la función mode() para obtener la moda de las variables categóricas
df_Lavoral['University_Rating'] = df_Lavoral['University_Rating'].fillna(df_Lavoral['University_Rating'].mode()[0])
```
